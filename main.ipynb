```python
from pathlib import Path
from rvseg.config import PathsConfig, PreprocConfig, PostprocConfig, ThresholdConfig, WeightSearchConfig, UncertaintyConfig
from rvseg.run_part1 import run_part1
from rvseg.run_part2 import run_part2
from rvseg.run_part3 import run_part3
from rvseg.run_part4 import run_part4
from rvseg.run_part5 import run_part5

paths = PathsConfig(
    images_dir=Path(r"..."), # insert images path
    masks_dir=Path(r"..."), # insert gt path
    masks2_dir=None # insert gt2 path if applicable
)

pre  = PreprocConfig(default_preproc="CLAHE_TOPHAT")
post = PostprocConfig(remove_small=True, min_obj_size=50)
thr  = ThresholdConfig(global_t=0.50)
ws   = WeightSearchConfig()
unc  = UncertaintyConfig(save_uncertainty_maps=True)

# Part 1: segmentation and analysis of best channel from 6 color space models
df_eval, sum_all, sum_def = run_part1(paths, pre, post, thr)

# Part 2: generate best weighted RGB channel
thresh_method, best_w = run_part2(paths, pre, post, thr, ws)

# Part 3: calculate uncertainties and convergence
df_unc, sum_unc = run_part3(paths, pre, post, thr, unc, best_w=best_w, thresh_method=thresh_method)

# Part 4: plot analysis
res4 = run_part4(paths, pre, post, thr, n_examples=3)

# Part 5: comparative statistical analysis of best channels
corr_df, dice_df, unc_df = run_part5()
```
